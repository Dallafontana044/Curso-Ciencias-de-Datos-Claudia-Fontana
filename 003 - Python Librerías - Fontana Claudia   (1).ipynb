```
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```
# crear un array con 5 numeros 
mis_numeros = np.array([10,20,30,40,50])
print (mis_numeros)
```

```
# crear una informacionde los estudiantes 
informacion= {
    'estudiantes':['Ana','juan','Carla','Pedro','Lucia'],
    'notas':[8,6,9,5,7]
}
df=pd.DataFrame(informacion)
print(df)    
```

```
#mostrar promedio,Maximo y minimo de las notas 
promedio=df['notas'].mean()
maximo=df['notas'].max()
minimo=df['notas'].min()
print(f"promedio de notas:{promedio}")
print(f"nota maxima:{maximo}")
print(f"nota minima:{minimo}") 
 
```
